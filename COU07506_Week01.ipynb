```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
```

```python

# TASK 1.1: GENERATING BASIC SIGNALS

# Create sample index
n = np.arange(-5, 11)

# Generate four signals
delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = (0.8**n) * step

# Plot the signals
fig, axes = plt.subplots(1, 4, figsize=(14, 3))

signals = [delta, step, ramp, expo]
titles = ["Unit Impulse", "Unit Step",
          "Unit Ramp", "Exponential"]

for ax, x, title in zip(axes, signals, titles):
    ax.stem(n, x)
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid()

plt.tight_layout()
plt.show()
```


### Observation – Task 1.1

The four basic discrete-time signals were successfully generated and plotted using Python.

1. Unit Impulse: Had a value of 1 at n = 0 and 0 elsewhere.
2. Unit Step: Changed from 0 to 1 at n = 0.
3. Unit Ramp: Increased linearly for positive values of n.
4. Exponential: Decreased as n increased because the value of a was 0.8.


```python

# TASK 1.2: EXPONENTIAL SIGNALS

n = np.arange(-5, 21)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(13, 3))

for ax, a in zip(axes, a_values):
    x = (a**n) * (n >= 0)
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid()

plt.tight_layout()
plt.show()
```


### Observation

- When a = 0.5, the exponential signal decreased.
- When a = -0.8, the signal alternated between positive and negative values while decreasing in magnitude.
- When a = 1.1, the exponential signal increased continuously.


```python

# COSINE SIGNALS

n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

fig, axes = plt.subplots(1, 2, figsize=(13, 3))

axes[0].stem(n, x1)
axes[0].set_title("cos(2πn/8)")
axes[0].set_xlabel("n (samples)")
axes[0].set_ylabel("Amplitude")
axes[0].grid()

axes[1].stem(n, x2)
axes[1].set_title("cos(2πn/7.5)")
axes[1].set_xlabel("n (samples)")
axes[1].set_ylabel("Amplitude")
axes[1].grid()

plt.tight_layout()
plt.show()
```


### Observation

The first cosine signal repeats every 8 samples, while the second cosine signal repeats every 15 samples. Both signals are periodic.
